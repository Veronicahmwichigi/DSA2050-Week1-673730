```python
import pandas as pd

#Loading the dataset
file=r"C:\Users\Elite X2 G4\OneDrive\Desktop\DSA2050\DSA_2050_Week1_Retail_Data.csv"
data=pd.read_csv(file)
```

```python
#Checking the shape and data types
print(data.head()) #displays the first 5 rows of the dataset
print(data.shape)  #data has 300 rows and 11 columns
print(data.info()) #3 columns have numeric variables(7-9)
                #and 8 columns have categorical variables (1-6, 10-11)
```

```python
#KPIs: 1. Top selling product 2.Revenue by category 3.Revenue by region
top_product=data.groupby("Product")["Quantity"].sum().sort_values(ascending=False)
print(top_product) #The top selling product is Laptop sleeve with a quantity of 60 while
                    #the least selling product is backpack with a quantity of 25

category_revenue=data.groupby("Category")["Revenue_KES"].sum().sort_values(ascending=False)
print(category_revenue) #The category with the highest revenue is Electronics with a Revenue of 934,500 while
                        #the category with the lowest revenue is Accessories with a Revenue of 422,250

region_revenue=data.groupby("Region")["Revenue_KES"].sum().sort_values(ascending=False)
print(region_revenue) #The region with the highest revenue is Kisumu with a Revenue of 878,050 while
                    #the region with the lowest revenue is Nairobi with a Revenue of 757,950
```

```python
revenue_channel=data.groupby("Channel")["Revenue_KES"].sum()
revenue_channel.plot(kind="bar", title="Revenue by Channel", xlabel="Channel", ylabel="Revenue (KES)") 
#The channel with the highest revenue is in-store with a revenue of 1.75 million while online has 1.35 million.
```

```python
payment_method=data.groupby("PaymentMethod")["Revenue_KES"].sum()
payment_method.plot(kind="bar", title="Revenue by Payment Method", xlabel="Payment Method", ylabel="Revenue (KES)")
#The payment method with the highest revenue is cash with a revenue of around 1.2 million while card has the lowest revenue of 1.0 million.
```
